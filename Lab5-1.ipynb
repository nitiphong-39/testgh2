#Lab5-1
Functions

**Why Use Functions?**

**ตัวอย่าง** โค้ดที่ไม่มีฟังก์ชัน

สังเกตว่าต้องเขียนสูตรเดิมซ้ำๆ ทุกครั้งที่มีสินค้าชิ้นใหม่ ถ้าภาษีเปลี่ยนจาก 7% เป็น 10% เราต้องตามแก้ทุกบรรทัด

```python
item1_price = 100
item1_total = item1_price + (item1_price * 0.07)
print(f"Item 1: {item1_total}")

item2_price = 250
item2_total = item2_price + (item2_price * 0.07)
print(f"Item 2: {item2_total}")

item3_price = 500
item3_total = item3_price + (item3_price * 0.07)
print(f"Item 3: {item3_total}")
```

**ตัวอย่าง** โค้ดที่มีการใช้ฟังก์ชัน

เขียนสูตรคำนวณไว้ที่เดียวในฟังก์ชันชื่อ net_price เมื่อต้องการใช้งานก็นำราคาไปใส่ในวงเล็บได้เลย

ถ้าอยากแก้ภาษีเป็น 10% แก้แค่บรรทัดที่ 2 จุดเดียวจบ

```python
def net_price(price):
    return price + (price * 0.07)

print(f"Item 1: {net_price(100)}")
print(f"Item 2: {net_price(250)}")
print(f"Item 3: {net_price(500)}")
```

**A function with one parameter**

**Number of parameter/argument**

จะมีกี่ตัวก็ได้ตามความต้องการที่จะใช้งาน

ตัวอย่างที่มากกว่า 1 Parameter

```python
def print_name(fname, lname):
  print(fname + " " + lname)


first_name = input("Enter First name:")
last_name = input("Enter Last name:")
print_name(first_name, last_name)
```

**Default Parameter Values**

หากมีการเรียกใช้งานฟังก์ชันโดยไม่มีการส่ง Parameter เข้าไป ฟังก์ชันจะนำค่าเริ่มต้นที่กำหนดไว้มาใช้โดยอัตโนมัติ

```python
def call_name(name = "Friend"):
  print("Hello", name)

call_name("Oil")
call_name("Jane")
call_name()
call_name("Bob")
```

```python
def print_country(country = "Thailand"):
  print("I am from", country)

print_country("Germany")
print_country()
```

**Keyword Arguments**
ใช้ key = value

```python
def print_animal(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

print_animal(animal = "cat", name = "MeeKeaw")
print_animal(name = "MeeKeaw", animal = "cat")
```

**Passing Different Data Types**

สามารถส่งข้อมูลประเภทใดก็ได้เข้าไปเป็น parameter ของฟังก์ชัน ไม่ว่าจะเป็น String, Number, List และอื่นๆ


ตัวอย่างการส่ง list

```python
def print_fruits(fruits):
  for fruit in fruits:
    print(fruit)

my_fruits = ["apple", "banana", "cherry"]
print_fruits(my_fruits)
```

**Return Values**

สามารถคืนค่าจากฟังก์ชั่นได้ด้วยคำสั่ง return

```python
def add(x, y):
  return x + y

result = add(5, 3)
print(result)
```

**Returning Different Data Types**

สามารถคืนค่าออกจากฟังก์ชัน ไม่ว่าจะเป็น String, Number, List และอื่นๆ


```python
def print_fruits():
  return ["apple", "banana", "cherry"]

fruits = print_fruits()
print(fruits[0])
print(fruits[1])
print(fruits[2])
```
