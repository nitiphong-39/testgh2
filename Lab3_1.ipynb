```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv2.imread("C:/Users/Home/Desktop/P2/imageprocess/dip_s2/Lab/imageEx/F309a.jpg", 0)

plt.imshow(img, cmap='gray')
```

```python
ret, img2 = cv2.threshold(img, 127,255, cv2.THRESH_BINARY)
plt.imshow(img2, cmap='gray')
```

```python
pattern = [cv2.THRESH_BINARY, cv2.THRESH_BINARY_INV, cv2.THRESH_TRUNC, cv2.THRESH_TOZERO, cv2.THRESH_TOZERO_INV]
title = ["THRESH_BINARY","THRESH_BINARY_INV","THRESH_TRUNC","THRESH_TOZERO","THRESH_TOZERO_INV"]
for i in range(5):
    ret, img2 = cv2.threshold(img, 200,205, pattern[i])
    plt.subplot(2,3,i+1)
    plt.title(title[i])
    plt.xticks([])
    plt.yticks([])
    plt.imshow(img2, cmap='gray')
plt.subplot(2,3,6)
plt.title(title[1])
plt.hist(img.ravel(),256,[0,256])
plt.show()
```
